```python
def even_or_odd(num):
    if num % 2 == 0:
        print("Even")
    else:
        print("Odd")
```

```python
#call this function
even_or_odd(10)  # Output: Even
```

```python
#multiple parameters
def add_numbers(a, b):
    return a + b

result = add_numbers(5, 7)
print(result)  # Output: 12
```

```python
#default parameters
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

greet("Alice")  # Output: Hello, Alice!
greet("Bob", "Hi")  # Output: Hi, Bob!
```

```python
#variable-length arguments
#positional and keyword variable-length arguments
def print_numbers(*args):
    for num in args:
        print(num)
print_numbers(1, 2, 3)
print()# Output: 1 2 3
print_numbers(4, 5)     # Output: 4 5
```

```python
#keyword args
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_details(name="Alice", age=30, city="New York")
```

```python
def print_details(*args,**kwargs):
    for number in args:
        print(f"positional argument: {number}")
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_details(1, 2, 3, name="Alice", age=30, city="New York")
```
