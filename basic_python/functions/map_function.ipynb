The map function applies a given function to all items in an input list. 
and returns a map object. This is particularly useful for transforming data in
a list comprehensively

```python
numbers = [1,2,3,4,5,6,7]
squared_numbers = list(map(lambda x: x**2, numbers))
print(squared_numbers)  # Output: [1, 4, 9, 16, 25, 36, 49]
```

```python
#map multiple iterables
numbers1 = [1,2,3]
numbers2 = [4,5,6]

added_numbers = list(map(lambda x,y: x+y, numbers1, numbers2))
print(added_numbers)  # Output: [5, 7, 9]
```

```python
#map() to convert a list of string to integers
string_numbers = ["1", "2", "3", "4", "5"]
int_numbers = list(map(int, string_numbers))
print(int_numbers)  # Output: [1, 2, 3, 4
```

```python
words = ["hello", "world", "python"]
upper_words = list(map(str.upper, words))
print(upper_words)  # Output: ['HELLO', 'WORLD', 'PYTHON']
```

```python
people = [
    {"name": "Alice", "age": 30},
    {"name": "Bob", "age": 25},
    {"name": "Charlie", "age": 35}]

names = list(map(lambda person: person["name"], people))
print(names)  # Output: ['Alice', 'Bob', 'Charlie']
```
