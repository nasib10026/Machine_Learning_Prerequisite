#Dictionaries
Dictionaries are unordered, changeable, and indexed collections of key-value pairs. Each key must be unique and immutable (such as strings, numbers, or tuples), while the values can be of any data type and can be duplicated. Dictionaries are defined using curly braces {} with key-value pairs separated by colons (:). They are commonly used for storing and retrieving data based on unique keys, making them a powerful tool for various applications in programming.

```python
#creating dictionaries
empty_dic = {}
print(type(empty_dic))
```

```python
empty_dict = dict()
print(type(empty_dict))
```

```python
student = {'name': 'Rakibul', 'age': 25, 'courses': ['Math', 'CompSci']}
print(student)
```

```python
print(type(student))
```

```python
#Error
# 2nd key will overwrite the first one. so unique keys are required in a dictionary
student = {'Name': 'Rakibul', 'age': 25, 'Name': 24}
print(student)
```

```python
student = {'name': 'Rakibul', 'age': 25, 'courses': ['Math', 'CompSci'],'grade': 'A'}
# accessing values from a dictionary
print(student['age'])
print(student['courses'])
print(student['courses'][0])
print(student['courses'][-1])  
print(student['grade'])

#Accessing using get method
print(student.get('name'))
print(student.get('phone'))# returns None if key is not found
print(student.get('phone', 'Not Available')) # returns 'Not Available' if key is not found
```

```python
#Modifying a dictionary elements. Dictionaries are mutable
student['age'] = 26
student['address'] = 'Dhaka' # adding new key-value pair
print(student)
```

```python
#delete a key, value pair from a dictionary
del student['grade']
```

```python
student
```

```python
#dictionary methods
keys = student.keys()
print(keys)
values = student.values()
print(values)
items = student.items()
print(items)
```

```python
#shallow copy of a dictionary
student_copy = student
print(student_copy)
print(student)
```

```python
student['name'] = 'Nasib 2'
print(student_copy)
print(student)
```

```python
student_copy1 = student.copy() # creates a shallow copy of the dictionary
print(student_copy1)
student['name'] = 'Nasib 3'
print(student_copy1)
print(student)
```

```python
##iterating over a dictionary
# try loops to iterate
for key in student.keys():
    print(key)
print()
for value in student.values():
    print(value)
print()
for key, value in student.items():
    print(f"{key}: {value}")
```

```python
##Nested dictionaries
students = {
    'student1': {'name': 'Rakibul', 'age': 25},
    'student2': {'name': 'Nasib', 'age': 24}
}
print(students)
```

```python
#Access nested dictionary values
print(students['student1']['name'])
print(students['student2']['age'])
```

```python
#Iterating over nested dictionaries
for student_id, student_info in students.items():
    print(f"{student_id}:")
    for key, value in student_info.items():
        print(f"  {key}: {value}")
```

```python
#Dictionary comprehension
squares = {x: x**2 for x in range(1, 6)}
squares
```

```python
#conditional dictionary comprehension
even_squares = {x: x**2 for x in range(1, 11) if x % 2 == 0}
even_squares
```

```python
#practical exaple
numbers = [1,1,1,3,4,5,5,7]
frequency = {num: numbers.count(num) for num in set(numbers)}
frequency
```

```python
# merge two dictionaries
dict1 = {'a': 1, 'b': 2}
dict2 = {'b': 3, 'c': 4}
merged_dict = {**dict1, **dict2}
print(merged_dict)
```
