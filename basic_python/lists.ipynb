#list comprehension

```python
lst = []

for x in range(1,11):
    lst.append(x**2)
print(lst)  # Output: [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]
```

```python
[x**2 for x in range(1,11)]  # Output: [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]
```

```python
#list comprehension with condition
lst = []
for x in range(1,11):
    if x % 2 == 0:
        lst.append(x**2)
print(lst)  # Output: [4, 16, 36, 64,
```

```python
[x**2 for x in range(1,11) if x%2 == 0]
```

```python
#nested list comprehension
lst1 = [1,2,3]
lst2 = ['a','b','c']

pair = [(x,y) for x in lst1 for y in lst2]
print(pair)
```

```python
#list comprehension with function calls
words = ["hello", "world", "python", "programming"]
lengths = [len(word) for word in words]
print(lengths) 
```
