```python
#File operations: Read and write files in Python
with open("example.txt","r") as file:
     content = file.read()
     print(content)
```

```python
#read a file line by line
with open("example.txt","r") as file:
    for line in file:
        print(line.strip())
```

```python
#Writing a file. overwriting
with open("example.txt",'w') as file:
    file.write("hello world\n")
    file.write("This is a new line\n")
     
    
```

```python
# write a file without overwriting it
with open("example.txt",'a') as file:
    file.write("Append operation taking place\n") 
```

```python
#writing list of lines in a file
lines = ["firstline\n",'secondline\n','thirdline\n']
with open("example.txt",'a') as file:
    file.writelines(lines)
    
```

```python
## Binary files
data = b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
    file.write(data)
```

```python
with open('example.bin','rb') as file:
    content = file.read()
    print(content)
```

```python
# read the content from a src text file and write it to a destination text file
#copying a text file
with open("example.txt",'r') as source_file:
    lines = source_file.readlines()

with open("example.txt",'r') as file:
    content = file.read()

with open("destination.text",'w') as dest_file:
    dest_file.write(content)
```

```python
#writing and then reading a file
with open("example.txt","w+") as file:
    file.write("Hello world\n")
    file.write("This is a new line\n") 
    
    ## move the file cursor to the beginning
    file.seek(0)
    
    ##read the content
    content = file.read()
    print(content)
    
```

```python
#create a new directory
import os
new_directory = "package"
os.mkdir(new_directory)
print(f"Directory '{new_directory} is created")
```

```python
#listing files and directories
items = os.listdir('.')
print(items)
```

```python
### joining paths
dir_name = "folder"
file_name = "file.txt"

full_path = os.path.join(os.getcwd(),dir_name,file_name)

print(full_path)
```

```python
path = 'example.txt'

if os.path.exists(path):
    print(f"Path '{path}' exists")
else:
    print("Path doesnt exist")
```

```python
#checking if a path is a file or directory
path = 'package'

if os.path.isfile(path):
    print(f"Path '{path}' is a file")
elif os.path.isdir(path):
    print(f"Path '{path}' is a folder")

else:
    print("Neither a file nor a directory")
```

```python
#Getting the absolute path
relative_path = "example.txt"
absolute_path = os.path.abspath(relative_path);
print(absolute_path)
```
