```python
##create a set
my_set = {1, 2, 3, 4, 5}
print(my_set) 
print(type(my_set))# Output: {1, 2, 3, 4
```

```python
my_empty_set = set()
print(type(my_empty_set))
```

```python
my_set = set([1, 2, 3, 4, 5,5,5,6,7])
print(my_set)
```

```python
##basic set operations
##adding and removing elements
my_set.add(8)
print(my_set)
#my_set.remove(3)
print(my_set)
#my_set.remove(10)#keyerror
my_set.discard(10) #no error
print(my_set)
```

```python
#pop method removes and returns an arbitrary element from the set. If the set is empty, it raises a KeyError.
popped_element = my_set.pop()
print(f"Popped element: {popped_element}")
print(my_set)
```

```python
my_set.pop()
print(my_set)
#first element gets removed 
#clear all elements from the set
my_set.clear()
print(my_set)
```

```python
##set membership test
my_set = {1, 2, 3, 4, 5}
print(3 in my_set)  # Output: True
print(6 in my_set)  # Output: False
```

```python
# mathematcal set operations
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6} 

##union
union_set = set_a.union(set_b)
print(f"Union: {union_set}")

##intersection
intersection_set = set_a.intersection(set_b)
print(f"Intersection: {intersection_set}")

set_a.intersection_update(set_b) #updates set_a with the intersection of set_a and set_b
print(f"Updated set_a after intersection_update: {set_a}")
```

```python
#difference
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
set_a.difference(set_b)
```

```python
#symmetric difference. unique elements in either set_a or set_b but not in both
set_a.symmetric_difference(set_b)
```

```python
#sets methods
set1 = {1, 2, 3,4,5}
set2 = {3, 4, 5}

##is subset
print(set1.issubset(set2))  # Output: False
print(set1.issubset({1, 2, 3, 4, 5}))  # Output: True
print(set1.issuperset(set2))  # Output: True
print(set1.issuperset({1, 2}))  # Output: True
```

```python
lst = [1,2,2,3,4,5,5,6]
set(lst)
```

```python
##counting unique elements in a list using sets
text = "In this sentence, we will count the number of unique words."
words = text.split()
unique_words = set(words)
print(f"Unique words: {unique_words}")
print(f"Number of unique words: {len(unique_words)}")
```
