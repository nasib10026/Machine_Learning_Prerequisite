```python
#creating a tuple
empty_tuple = ()
print(empty_tuple)  # Output: ()
print(type(empty_tuple))  # Output: <class 'tuple'>
```

```python
lst = list()
print(lst)  # Output: []
print(type(lst))  # Output: <class 'list'>
tpl = tuple()
print(tpl)  # Output: ()
print(type(tpl))  # Output: <class 'tuple'>
```

```python
numbers = tuple([1, 2, 3, 4, 5])
print(numbers)  # Output: (1, 2, 3, 4,
print(list(numbers))  # Output: [1, 2, 3, 4, 5]
```

```python
#mixed tuple
mixed_tuple = (1, "hello", 3.14, [1, 2, 3], (4, 5))
print(mixed_tuple)  # Output: (1, 'hello', 3.14
```

```python
#accessing tuple elements
numbers
 
```

```python
print(numbers[0])  # Output: 1
print(numbers[-1])  # Output: 2
```

```python
#slicing will be similar to list slicing
print(numbers[1:4])
print(numbers[::-1])# Output: (2, 3, 4
```

```python
#tuple_operations
concatenated_tuple = numbers + mixed_tuple
print(concatenated_tuple)  # Output: (1, 2, 3
```

```python
mixed_tuple_repeated = mixed_tuple * 2
print(mixed_tuple_repeated)  # Output: (1, 'hello', 3
```

```python
#immutable nature of tuples
#tuples are immutable, which means you cannot change their elements after creation
lst = [1, 2, 3, 4, 5]
print(lst)
lst[0] = 'Nasib'
print(lst)  # Output: ['Nasib', 2, 3, 4
```

```python
#in case of tuple, it will raise an error
tpl = (1, 2, 3, 4, 5)
print(tpl)
tpl[0] = 'Nasib'  # This will raise a TypeError: '
print(tpl)
```

```python
#Tuple methods
print(numbers.count(1)) 
# 3 present in second index
print(numbers.index(3))# Output: 1
```

```python
#packing and unpacking tuples
#packing
packed_tuple = 1, 'hello', 3.14
print(packed_tuple)  # Output: (1, 'hello', 3.14)
#unpacking
a, b ,c= packed_tuple
print(a)  # Output: 1
print(b)  # Output: 'hello'
  # Output: 3.14
```

```python
#unpacking with *
numbers = (1, 2, 3, 4, 5)
furst, *middle, last = numbers
print(furst)  # Output: 1
print(middle)  # Output: [2, 3, 4]
print(last)  # Output: 5
```

```python
#nested tuples
#nested list
lst = [[1,2],[3,4],['ArithmeticError', 'ValueError']]
print(lst)  # Output: [[1, 2], [3, 4], ['ArithmeticError', 'ValueError']]
print(lst[2][0:2])
```

```python
#nested tuples
nested_tuple = ((1,2), (2, 3), (4, 5, 6))
print(nested_tuple)  # Output: (1, (2, 3), (4, 5, 6))
print(nested_tuple[1])  # Output: (2, 3)
print(nested_tuple[2][1])  # Output:
```

```python
#iterating over a nested tuple using for loop
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end = " ")
    print()
```

###conclusion
Tuples are immutable, ordered collections that can contain heterogeneous data types. They support various operations such as indexing, slicing, concatenation, and repetition. Tuples also have methods like count() and index() for counting occurrences and finding the index of elements. Additionally, tuples can be nested and unpacked using packing and unpacking techniques.
